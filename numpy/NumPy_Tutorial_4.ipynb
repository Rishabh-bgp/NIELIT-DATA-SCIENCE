# **1. Sorting One Array Based on Another (Linked Data)**

```python
names = np.array(["Alice", "Bob", "Charlie", "David"])
scores = np.array([85, 92, 78, 95])

sort_idx = np.argsort(scores)

sorted_names = names[sort_idx]
print(sorted_names)
```

# **2. Finding Top-N or Bottom-N Elements**

```python
prices = np.array([120, 45, 300, 15, 80])

sorted_idx = np.argsort(prices)

cheapest_idx = sorted_idx[:2]

expensive_idx = sorted_idx[-2:]
```

# **3. Rearranging Rows in a Matrix/DataFrame**

```python
matrix = np.array([
    [101, 35, 50000],
    [102, 22, 45000],
    [103, 29, 60000]
])

age_idx = np.argsort(matrix[:, 1])

sorted_matrix = matrix[age_idx]
print("Original matrix:\n", matrix)
print("\nSorted matrix by age (column 1):\n", sorted_matrix)
```
