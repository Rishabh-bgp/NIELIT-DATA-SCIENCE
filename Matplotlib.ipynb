# Introduction to Matplotlib

`matplotlib` is one of the most widely-used data visualization libraries in Python. It provides a flexible and powerful platform for creating static, animated, and interactive visualizations.

### Key Concepts:
- **Figure (`fig`)**: The overall window or page where everything is drawn. It can contain one or more plots (axes).
- **Axes (`ax`)**: The actual plot (with x and y axes, ticks, labels, and title). A single Figure can contain multiple Axes.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

print("Libraries imported and data prepared.")
```

## 1. Creating a Basic Line Plot

Let's start by plotting a simple sine wave using `plt.plot()`.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y1, label='Sine Wave', color='blue', linewidth=2)
plt.title('Basic Sine Wave Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

## 2. Object-Oriented Interface (Recommended)

While `plt.plot()` is quick, the Object-Oriented (OO) style is more robust and recommended for complex layouts. Here we explicitly create a `Figure` and `Axes` object.

```python
fig, ax = plt.subplots(figsize=(8, 4))

# Plotting multiple lines on the same axes
ax.plot(x, y1, label='Sine', color='teal', linestyle='-')
ax.plot(x, y2, label='Cosine', color='coral', linestyle='--')

# Customizing using axes methods
ax.set_title('Sine and Cosine Waves', fontsize=14, fontweight='bold')
ax.set_xlabel('Time', fontsize=12)
ax.set_ylabel('Amplitude', fontsize=12)
ax.grid(True)
ax.legend(loc='upper right')

plt.show()
```

## 3. Subplots: Multiple Plots in One Figure

We can create grid layouts using `plt.subplots(rows, cols)`.

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Left subplot
axes[0].plot(x, y1, color='green')
axes[0].set_title('Left Subplot (Sine)')
axes[0].grid(True)

# Right subplot
axes[1].plot(x, y2, color='red')
axes[1].set_title('Right Subplot (Cosine)')
axes[1].grid(True)

plt.tight_layout() # Adjust spacing
plt.show()
```

## 4. Other Plot Types

Matplotlib supports a variety of plots like Scatter, Bar, and Histogram plots.

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Scatter Plot
categories = np.random.rand(50)
colors = np.random.rand(50)
axes[0].scatter(np.random.rand(50), np.random.rand(50), c=colors, s=categories*100, alpha=0.7, cmap='viridis')
axes[0].set_title('Scatter Plot')

# 2. Bar Plot
labels = ['A', 'B', 'C', 'D']
values = [3, 12, 5, 18]
axes[1].bar(labels, values, color='skyblue', edgecolor='black')
axes[1].set_title('Bar Chart')

# 3. Histogram
data = np.random.randn(1000)
axes[2].hist(data, bins=30, color='purple', edgecolor='white', alpha=0.7)
axes[2].set_title('Histogram')

plt.tight_layout()
plt.show()
```
