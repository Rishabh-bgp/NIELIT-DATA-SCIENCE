```python
import pandas as pd
import numpy as np

data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'Finance'],
    'Base_Salary': [60000, 95000, 70000, 110000, 85000],
    'Performance_Score': [4.2, 3.8, 4.9, 4.5, 3.9]
}

df = pd.DataFrame(data)
df.set_index('Employee_ID', inplace=True)

print("--- INITIAL DATAFRAME ---")
print(df, "\n" + "="*60 + "\n")

print("--- 1.1 Column Slicing ---")
names_series = df['Name']

dept_salaries_df = df[['Department', 'Base_Salary']]
print(dept_salaries_df, "\n")


print("--- 1.2 Label-Based Slicing (.loc) ---")
salary_slice = df.loc[101:103, 'Base_Salary']
print(salary_slice, "\n")


print("--- 1.3 Position-Based Slicing (.iloc) ---")
integer_slice = df.iloc[0:3, 0:2]
print(integer_slice, "\n" + "="*60 + "\n")

print("--- 2.1 Direct Assignment (In-Place Modifications) ---")
df['Is_Active'] = True

df['Location'] = ['New York', 'San Francisco', 'Chicago', 'Austin', 'Boston']
print(df[['Name', 'Is_Active', 'Location']], "\n")


print("--- 2.2 Pure Functional Copy Generation (.assign) ---")
df_expanded = df.assign(
    Bonus_Amount = df['Base_Salary'] * 0.10,
    Total_Compensation = lambda x: x['Base_Salary'] + (x['Base_Salary'] * 0.10)
)
print("Functional Copy Subset (df_expanded):")
print(df_expanded[['Name', 'Base_Salary', 'Bonus_Amount', 'Total_Compensation']], "\n")


print("--- 2.3 Explicit Position Placement (.insert) ---")
emails = ['alice@co.com', 'bob@co.com', 'charlie@co.com', 'david@co.com', 'eva@co.com']
df.insert(1, 'Corporate_Email', emails)
print(df, "\n")


print("--- 2.4 Conditional/Rule-Based Allocation ---")
df['Tier'] = 'Junior'
df.loc[df['Base_Salary'] > 90000, 'Tier'] = 'Senior'

rating_conditions = [
    (df['Performance_Score'] >= 4.5),
    (df['Performance_Score'] >= 4.0) & (df['Performance_Score'] < 4.5),
    (df['Performance_Score'] < 4.0)
]
rating_outputs = ['Excellent', 'Good', 'Satisfactory']

df['Performance_Rating'] = np.select(rating_conditions, rating_outputs, default='Unrated')

print("--- FINAL PROCESSED DATAFRAME ---")
print(df)
```
