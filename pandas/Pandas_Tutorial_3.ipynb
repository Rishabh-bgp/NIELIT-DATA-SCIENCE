# Titanic Dataset Preprocessing
This notebook demonstrates how to load the Titanic dataset and perform common preprocessing steps using `pandas`.

```python
import pandas as pd

# Load the Titanic dataset from an online repository
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

# Display the first few rows of the dataset
display(df.head())
```

## Data Exploration and Handling Missing Values
Let's check the dataset structure, data types, and count missing values in each column.

```python
# Print dataset summary
print("Dataset Summary:")
print(df.info())

# Count missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Handle missing values
# 1. Fill missing 'Age' with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# 2. Fill missing 'Embarked' with the most common value (mode)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# 3. Drop the 'Cabin' column due to a high ratio of missing values
if 'Cabin' in df.columns:
    df = df.drop(columns=['Cabin'])

# Verify that there are no more critical missing values left
print("Remaining missing values:")
print(df.isnull().sum())
```

## Feature Encoding & Engineering
We will encode categorical features (`Sex` and `Embarked`) into numerical values and drop unneeded columns like `Name`, `Ticket`, and `PassengerId`.

```python
# Encode categorical features using one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)

# Drop columns that aren't useful for general predictive modeling
columns_to_drop = ['PassengerId', 'Name', 'Ticket']
df_cleaned = df_encoded.drop(columns=[col for col in columns_to_drop if col in df_encoded.columns])

# Display the preprocessed dataset
display(df_cleaned.head())
```

## Exporting the Preprocessed Data

To save your cleaned DataFrame back to a CSV file, you can use the pandas `to_csv()` method. By default, this method includes the DataFrame index as a column. To prevent this, you can set `index=False`.

```python
# Export the cleaned DataFrame to a CSV file
output_filename = 'titanic_preprocessed.csv'
df_cleaned.to_csv(output_filename, index=False)

print(f"Successfully saved preprocessed data to '{output_filename}'!")
```
